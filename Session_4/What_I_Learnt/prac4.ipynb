# 📘 Session 4 Practice: Concurrency & Multithreading in Python

This notebook illustrates background thread execution using Python's standard `threading` module. Concurrent execution patterns are foundational when running streaming pipelines, handling asynchronous tool callbacks, or processing background tasks without blocking the user interface.

## 1. Imports
Import `threading` for background worker orchestration and `time` to simulate IO/network delays.

```python
import threading
import time
```

## 2. Concurrent Background Worker Execution
Spawn a background daemon thread that produces outputs at 5-second intervals while the main thread simultaneously continues executing its own sequence at 1-second intervals.

```python
def print_numbers() -> None:
    """Background worker function simulating periodic IO/telemetry tasks."""
    for i in range(1, 6):
        print(f"[Background Thread] Number: {i}")
        time.sleep(5)

# Start worker on separate thread
t = threading.Thread(target=print_numbers)
t.start()

# Main thread continues execution concurrently
for i in range(5):
    print(f"[Main Thread] Processing step {i + 1}")
    time.sleep(1)
```
